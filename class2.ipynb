```python
array=np.linspace(12,22,num=15, dtype=int)
```

```python
def variance (*args):
  total=0
  for i in args:
    total+=(i-np.mean(args))**2
  calculated_variance= total/(len(args))
  return calculated_variance
```

```python
variance(*array)
```

```python
def variance_std (*args):
  total=0
  for i in args:
    total+=(i-np.mean(args))**2
  calculated_variance= total/(len(args))
  return calculated_variance, calculated_variance**(1/2)
```

```python
variance_std (*array)
```

```python
array
```

```python
def sample_variance (*args):
  total=0
  for i in args:
    total+=(i-np.mean(args))**2
  cal_sample= total/(len(args)-1)
  return cal_sample
```

```python
sample_variance(*array)
```

```python
def sample_variance_std (*args):
  total=0
  for i in args:
    total+=(i-np.mean(args))**2
  cal_sample= total/(len(args)-1)
  return cal_sample, cal_sample**(1/2)
```

```python
sample_variance_std(*array)
```
